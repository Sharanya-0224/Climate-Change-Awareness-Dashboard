```python

# Plot annual climate drivers, distinguishing observed anchors
# from synthetic CH4 and N2O estimates.

variables = {
    "co2": ("CO₂ concentration", "ppm"),
    "sunspots": ("Sunspot number", "count"),
    "nino34": ("Niño 3.4 index", "°C"),
}

for col, (title, unit) in variables.items():
    fig, ax = plt.subplots(figsize=(12, 4.5))
    ax.plot(annual["year"], annual[col], linewidth=1.3)
    ax.set_title(f"{title} over Time")
    ax.set_xlabel("Year")
    ax.set_ylabel(unit)
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    fig.savefig(REPORT_DIR / f"{col}_annual_timeseries.png",
                dpi=150, bbox_inches="tight")
    plt.show()
    plt.close(fig)


# CH4 and N2O: display reconstructed series and mark anchors.
for gas, label, unit in [
    ("ch4", "Methane (CH₄)", "ppb"),
    ("n2o", "Nitrous oxide (N₂O)", "ppb"),
]:
    fig, ax = plt.subplots(figsize=(12, 5))

    ax.plot(
        annual["year"], annual[gas],
        linewidth=1.5, label="Reconstructed series"
    )

    synthetic_col = f"{gas}_is_synthetic"
    synthetic = annual[synthetic_col].fillna(False).astype(bool)

    # Non-synthetic values are plotted separately as anchor points.
    observed = annual[gas].notna() & ~synthetic
    estimated = annual[gas].notna() & synthetic

    ax.scatter(
        annual.loc[observed, "year"],
        annual.loc[observed, gas],
        s=22, label="Non-synthetic / anchor values", zorder=3
    )

    ax.scatter(
        annual.loc[estimated, "year"],
        annual.loc[estimated, gas],
        s=12, label="Synthetic estimates", zorder=2
    )

    ax.set_title(f"{label}: Observed Anchors vs Estimates")
    ax.set_xlabel("Year")
    ax.set_ylabel(unit)
    ax.grid(True, alpha=0.3)
    ax.legend()
    plt.tight_layout()
    fig.savefig(
        REPORT_DIR / f"{gas}_annual_observed_vs_estimated.png",
        dpi=150, bbox_inches="tight"
    )
    plt.show()
    plt.close(fig)
```

```python
import matplotlib.pyplot as plt

# Approximate CO2 radiative forcing relative to 280 ppm
monthly["co2_forcing"] = (
    5.35 * np.log(monthly["co2_trend"] / 280)
)

# Compare past CO2 forcing with temperature at the current month.
# A lag of 60 means forcing from 60 months earlier.
from scipy.stats import pearsonr

results = []

for lag in range(0, 121):  # 0 to 120 months (10 years)
    past_forcing = monthly["co2_forcing"].shift(lag)
    valid = pd.concat(
        [monthly["temp_anomaly"], past_forcing], axis=1
    ).dropna()

    if len(valid) >= 30:
        r, p = pearsonr(valid.iloc[:, 0], valid.iloc[:, 1])
        results.append({
            "lag_months": lag,
            "lag_years": lag / 12,
            "correlation": r,
            "p_value": p,
            "n": len(valid)
        })

lag_results = pd.DataFrame(results)

plt.figure(figsize=(12, 5))
plt.plot(lag_results["lag_years"], lag_results["correlation"])
plt.axhline(0, linewidth=1)
plt.xlabel("CO₂ forcing lag (years)")
plt.ylabel("Pearson correlation")
plt.title("Monthly temperature vs. earlier CO₂ forcing")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

display(
    lag_results.loc[
        lag_results["correlation"].abs().nlargest(10).index
    ].sort_values("lag_months")
)
```

```python
# CELL 1: RAW VS PROCESSED TEMPERATURE

raw_dir = PROJECT_ROOT / "data" / "raw"

raw_temp = pd.read_csv(
    raw_dir / "GLB.Ts+dSST.csv",
    skiprows=1,
    na_values=["***"]
)

raw_annual = (
    raw_temp[["Year", "J-D"]]
    .rename(columns={"Year": "year", "J-D": "temp_raw"})
    .dropna(subset=["temp_raw"])
    .copy()
)

raw_annual["year"] = raw_annual["year"].astype(int)

processed_annual = pd.read_csv(
    DATA_DIR / "climate_annual.csv"
)[["year", "temp_anomaly"]]

comparison = raw_annual.merge(
    processed_annual,
    on="year",
    how="inner",
    validate="one_to_one"
)

comparison["absolute_difference"] = (
    comparison["temp_raw"] - comparison["temp_anomaly"]
).abs()

print("Raw annual rows:", len(raw_annual))
print("Processed annual rows:", len(processed_annual))
print("Years compared:", len(comparison))
print(
    "Maximum absolute difference:",
    comparison["absolute_difference"].max()
)

display(comparison.head())

fig, ax = plt.subplots(figsize=(13, 5))

ax.plot(
    raw_annual["year"], raw_annual["temp_raw"],
    label="Raw NASA GISTEMP J-D", linewidth=1.5
)

ax.plot(
    processed_annual["year"], processed_annual["temp_anomaly"],
    linestyle="--", label="Processed annual temperature",
    linewidth=1.2
)

ax.axhline(0, color="black", linestyle=":", linewidth=1)
ax.set_title("Raw vs Processed Annual Temperature Anomaly")
ax.set_xlabel("Year")
ax.set_ylabel("Temperature anomaly (°C)")
ax.grid(alpha=0.25)
ax.legend()
plt.tight_layout()

fig.savefig(
    REPORT_DIR / "01_raw_vs_processed_temperature.png",
    dpi=180, bbox_inches="tight"
)
plt.show()
plt.close(fig)

comparison.to_csv(
    REPORT_DIR / "raw_vs_processed_temperature_check.csv",
    index=False
)
```

```python
# CELL 2: MISSINGNESS AND RECONSTRUCTION AUDIT

original = pd.read_csv(DATA_DIR / "climate_annual.csv")
reconstructed = pd.read_csv(
    DATA_DIR / "climate_annual_reconstructed.csv"
)
monthly = pd.read_csv(DATA_DIR / "climate_monthly.csv")
drivers_original = pd.read_csv(
    DATA_DIR / "drivers_annual_full.csv"
)
drivers_reconstructed = pd.read_csv(
    DATA_DIR / "drivers_annual_full_reconstructed.csv"
)

audit_rows = []

for dataset_name, df in [
    ("Annual original", original),
    ("Annual reconstructed", reconstructed),
    ("Full drivers original", drivers_original),
    ("Full drivers reconstructed", drivers_reconstructed),
    ("Monthly original", monthly)
]:
    for col in df.columns:
        if col in [
            "temp_anomaly", "co2", "co2_monthly", "co2_trend",
            "sunspots", "nino34", "ch4", "ch4_trend", "n2o", "n2o_trend"
        ]:
            audit_rows.append({
                "dataset": dataset_name,
                "variable": col,
                "rows": len(df),
                "missing_count": int(df[col].isna().sum()),
                "missing_percent": round(
                    df[col].isna().mean() * 100, 2
                ),
                "available_count": int(df[col].notna().sum())
            })

missing_audit_all = pd.DataFrame(audit_rows)

display(
    missing_audit_all[
        missing_audit_all["variable"].isin(
            ["temp_anomaly", "co2", "co2_monthly",
             "nino34", "ch4", "n2o"]
        )
    ]
)

missing_audit_all.to_csv(
    REPORT_DIR / "02_missingness_before_after.csv",
    index=False
)

# Focused CH4 / N2O comparison for annual climate data
gas_audit = pd.DataFrame({
    "CH4 missing": [
        original["ch4"].isna().sum(),
        reconstructed["ch4"].isna().sum()
    ],
    "N2O missing": [
        original["n2o"].isna().sum(),
        reconstructed["n2o"].isna().sum()
    ]
}, index=["Original", "Reconstructed"])

display(gas_audit)

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

gas_audit["CH4 missing"].plot(
    kind="bar", ax=axes[0], title="Missing CH₄ Values"
)
gas_audit["N2O missing"].plot(
    kind="bar", ax=axes[1], title="Missing N₂O Values"
)

for ax in axes:
    ax.set_ylabel("Number of missing annual values")
    ax.tick_params(axis="x", rotation=0)
    ax.grid(axis="y", alpha=0.25)

plt.suptitle("Effect of Annual GHG Reconstruction")
plt.tight_layout()
fig.savefig(
    REPORT_DIR / "03_missingness_before_after_reconstruction.png",
    dpi=180, bbox_inches="tight"
)
plt.show()
plt.close(fig)
```

```python
# CELL 3: OBSERVED ANCHORS VS SYNTHETIC ESTIMATES

for gas, label, unit in [
    ("ch4", "Methane (CH₄)", "ppb"),
    ("n2o", "Nitrous oxide (N₂O)", "ppb")
]:
    flag = f"{gas}_is_synthetic"

    # Fail clearly if the expected audit flag is absent
    if flag not in reconstructed.columns:
        raise KeyError(f"Missing reconstruction flag: {flag}")

    is_synthetic = (
        reconstructed[flag].fillna(False).astype(bool)
    )

    observed = (
        reconstructed[gas].notna() & ~is_synthetic
    )
    estimated = (
        reconstructed[gas].notna() & is_synthetic
    )

    fig, ax = plt.subplots(figsize=(13, 5))

    ax.plot(
        reconstructed["year"], reconstructed[gas],
        color="gray", linewidth=1.2,
        label="Reconstructed series"
    )

    ax.scatter(
        reconstructed.loc[observed, "year"],
        reconstructed.loc[observed, gas],
        s=25, label="Non-synthetic anchors", zorder=3
    )

    ax.scatter(
        reconstructed.loc[estimated, "year"],
        reconstructed.loc[estimated, gas],
        s=15, label="Synthetic estimates", zorder=2
    )

    ax.set_title(f"{label}: Observed Anchors and Reconstructed Values")
    ax.set_xlabel("Year")
    ax.set_ylabel(f"Concentration ({unit})")
    ax.grid(alpha=0.25)
    ax.legend()
    plt.tight_layout()

    fig.savefig(
        REPORT_DIR / f"04_{gas}_observed_vs_reconstructed.png",
        dpi=180, bbox_inches="tight"
    )
    plt.show()
    plt.close(fig)

    print(f"\n{label}")
    print("Observed/non-synthetic rows:", int(observed.sum()))
    print("Synthetic rows:", int(estimated.sum()))
    print("Missing rows:", int(reconstructed[gas].isna().sum()))
    print("\nStatus counts:")
    display(reconstructed[f"{gas}_data_status"].value_counts(dropna=False))
```

```python
# CELL 6: CORRELATIONS WITH PAIRWISE SAMPLE COUNTS

corr_cols = [
    "temp_anomaly", "co2", "sunspots",
    "nino34", "ch4", "n2o"
]

def correlation_and_counts(df, cols):
    data = df[cols].apply(pd.to_numeric, errors="coerce")
    corr = data.corr()
    counts = data.notna().astype(int).T.dot(
        data.notna().astype(int)
    )
    return corr, counts

def plot_corr_with_counts(corr, counts, title, filename):
    fig, ax = plt.subplots(figsize=(10, 8))

    im = ax.imshow(corr, vmin=-1, vmax=1, cmap="coolwarm")
    ax.set_xticks(range(len(corr.columns)))
    ax.set_yticks(range(len(corr.index)))
    ax.set_xticklabels(corr.columns, rotation=45, ha="right")
    ax.set_yticklabels(corr.index)

    for i in range(len(corr.index)):
        for j in range(len(corr.columns)):
            r = corr.iloc[i, j]
            n = counts.iloc[i, j]
            label = "N/A" if pd.isna(r) else f"{r:.2f}\nn={n}"
            ax.text(j, i, label, ha="center", va="center", fontsize=8)

    fig.colorbar(im, ax=ax, label="Pearson correlation")
    ax.set_title(title)
    plt.tight_layout()
    fig.savefig(REPORT_DIR / filename, dpi=180, bbox_inches="tight")
    plt.show()
    plt.close(fig)

# Original measured data: missing GHG values stay missing
observed_corr, observed_n = correlation_and_counts(
    original, corr_cols
)
plot_corr_with_counts(
    observed_corr, observed_n,
    "Annual Correlations — Original Data (Pairwise n shown)",
    "07_correlation_original_observed.png"
)

# Reconstructed data: explicitly exploratory, not observed-only
recon_corr, recon_n = correlation_and_counts(
    reconstructed, corr_cols
)
plot_corr_with_counts(
    recon_corr, recon_n,
    "Annual Correlations — Reconstructed Series (Pairwise n shown)",
    "08_correlation_reconstructed_exploratory.png"
)

observed_corr.to_csv(REPORT_DIR / "correlation_original_observed.csv")
observed_n.to_csv(REPORT_DIR / "correlation_original_pair_counts.csv")
recon_corr.to_csv(REPORT_DIR / "correlation_reconstructed.csv")
recon_n.to_csv(REPORT_DIR / "correlation_reconstructed_pair_counts.csv")

print("Observed-only correlation counts:")
display(observed_n)
```

```python
# CELL 7: YEAR-TO-YEAR CHANGE CORRELATIONS — ORIGINAL DATA

change_cols = [
    "temp_anomaly", "co2", "sunspots",
    "nino34", "ch4", "n2o"
]

observed_sorted = original.sort_values("year").copy()

# Important: diff() is performed before dropping missing rows.
# We must not bridge across gaps in the original time series.
annual_changes = observed_sorted[change_cols].diff()

change_corr, change_n = correlation_and_counts(
    annual_changes, change_cols
)

plot_corr_with_counts(
    change_corr, change_n,
    "Annual Change Correlations — Original Observations",
    "09_annual_change_correlation_observed.png"
)

change_corr.to_csv(
    REPORT_DIR / "annual_change_correlation_observed.csv"
)
change_n.to_csv(
    REPORT_DIR / "annual_change_correlation_counts.csv"
)

display(change_n)
```

```python
# CELL 8: MONTHLY CO2-FORCING LAG DIAGNOSTICS

from scipy.stats import pearsonr
from scipy.signal import detrend

monthly = pd.read_csv(DATA_DIR / "climate_monthly.csv")
monthly["date"] = pd.to_datetime(monthly["date"])
monthly = monthly.sort_values("date").reset_index(drop=True)

# Approximate radiative forcing from deseasonalized CO2.
if (monthly["co2_trend"].dropna() <= 0).any():
    raise ValueError("CO2 trend contains non-positive values.")

monthly["co2_forcing"] = (
    5.35 * np.log(monthly["co2_trend"] / 280)
)

# Use a common complete series before detrending.
valid = monthly[["date", "temp_anomaly", "co2_forcing"]].dropna().copy()

valid["temp_detrended"] = detrend(
    valid["temp_anomaly"].to_numpy()
)
valid["forcing_detrended"] = detrend(
    valid["co2_forcing"].to_numpy()
)

results = []

for lag in range(0, 121):
    # At lag 60, forcing from 60 months earlier is paired
    # with temperature at the current point.
    for mode, temp_col, forcing_col in [
        ("Levels", "temp_anomaly", "co2_forcing"),
        ("Linearly detrended", "temp_detrended", "forcing_detrended")
    ]:
        x = valid[forcing_col].shift(lag)
        y = valid[temp_col]

        pair = pd.concat([y, x], axis=1).dropna()

        if len(pair) >= 100:
            r, p = pearsonr(pair.iloc[:, 0], pair.iloc[:, 1])
            results.append({
                "mode": mode,
                "lag_months": lag,
                "lag_years": lag / 12,
                "correlation": r,
                "p_value_naive": p,
                "n": len(pair)
            })

lag_results = pd.DataFrame(results)
lag_results.to_csv(
    REPORT_DIR / "10_monthly_co2_forcing_lag_results.csv",
    index=False
)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.8), sharey=True)

for ax, mode in zip(axes, ["Levels", "Linearly detrended"]):
    subset = lag_results[lag_results["mode"] == mode]

    ax.plot(subset["lag_years"], subset["correlation"])
    ax.axhline(0, color="black", linewidth=1, linestyle=":")
    ax.set_title(mode)
    ax.set_xlabel("Forcing lag (years)")
    ax.grid(alpha=0.25)

axes[0].set_ylabel("Pearson correlation")
fig.suptitle("Monthly Temperature vs Earlier CO₂ Forcing")
plt.tight_layout()

fig.savefig(
    REPORT_DIR / "10_monthly_co2_forcing_lag_diagnostic.png",
    dpi=180, bbox_inches="tight"
)
plt.show()
plt.close(fig)

# Show the highest positive and negative correlations in each mode.
for mode in ["Levels", "Linearly detrended"]:
    subset = lag_results[lag_results["mode"] == mode]

    print(f"\n{mode}: largest absolute correlations")
    display(
        subset.loc[
            subset["correlation"].abs().nlargest(10).index
        ].sort_values("lag_months")
    )
```
